# 2.3. Linear Algebra

Linear algebra is an indispensable tool in machine learning. Let's dive in!

```python
import mindspore
mindspore.set_device(device_target='Ascend', device_id=0)
```

## 2.3.1. Scalars

Numbers are sometimes known as scalars. They are tensors that contain only a single element.

```python
x = mindspore.Tensor(3.0)
y = mindspore.Tensor(2.0)
x + y, x - y, x * y, x / y, x ** y
```

## 2.3.2. Vectors

Vectors are tensors with order 1, otherwise known as first-order tensors. They are like lists or arrays of a fixed size.

```python
import mindspore.ops as ops
x = ops.arange(3)
x, x.size, x.shape
```

We can access elements in a vector by indexing just like we do a Python list.

```python
x[2]
```

The length of a vector is its size, i.e. number of elements.

```python
len(x)
```

## 2.3.3. Matrices

A matrix is a 2nd-order tensor, similar to how:

1. Scalars are zero-order tensors
1. Vectors are 1st-order tensors

We often use capital letters to denote them.

```python
A = ops.reshape(ops.arange(6), (3, 2))
A, A.size, A.shape
```

One common operation on matrices is taking its transpose. The transpose of a matrix simply swaps rows with columns and vice versa.

```python
ops.transpose(A, dims=(1, 0))
```

Square matrices have the same number of rows and columns. Furthermore, symmetric square matrices satisfy the property $A^{T} = A$.

```python
A = mindspore.Tensor([[1, 2, 3], [2, 0, 4], [3, 4, 5]])
ops.all(A == ops.transpose(A, dims=(1, 0))).item()
```

## 2.3.4. Tensors

Tensors are nth-order arrays determined by the size of their `shape` attribute. Common use cases of higher-order tensors include:

1. 3rd-order tensor: images \(height x width x _channel_ \(RGB\)\)
1. 4th-order tensor: collections of images \(number of images x height x width x channel \(RGB\)\)

```python
X = ops.reshape(ops.arange(24), (2, 3, 4))
X, X.size, X.shape, len(X.shape)
```

## 2.3.5. Basic Properties of Tensor Arithmetic

Addition works element-wise.

```python
A = ops.reshape(ops.arange(6), (2, 3))
B = A.clone()
A, B, A + B
```

Element-wise multiplication is denoted by the Python `*` operator. It's also known as the Hadamard product.

```python
A * B
```

Adding or multiplying a scalar with a tensor produces a new tensor with the same shape of the input tensor. The scalar and operation is applied to each element in the tensor.

```python
B[:] = 1 + A
C = 2 * A
B, C, A.shape == B.shape, A.shape == C.shape
```

## 2.3.6. Reduction

We can sum the elements in a tensor with the `ops.sum` method.

```python
ops.sum(A).item()
```

Similarly, use the `ops.prod` method to compute the product of all elements in a tensor.

```python
ops.prod(B).item()
```

We can reduce a tensor along a specific dimension as well by specifying the `dim` parameter.

```python
d = ops.sum(C, dim=0)
f = ops.sum(C, dim=1)
d, f
```

Reducing a matrix along both rows and columns is identical to omitting the `dim` parameter in `ops.sum`, or `axis` parameter in `ops.prod`.

```python
ops.prod(B, axis=(0, 1)).item()
```

We can also compute the mean average defined as the sum of all elements divided by the total number of elements.

```python
from mindspore import dtype as mstype
G = ops.cast(A, dtype=mstype.float)
ops.mean(G).item()
```

Compute the mean along a specific dimension with the `axis` parameter.

```python
ops.mean(G, axis=0), ops.mean(G, axis=1)
```

```python
ops.all(ops.mean(G, axis=1) == ops.sum(G, dim=1) / G.shape[1]).item()
```

## 2.3.7. Non-Reduction Sum

When computing the sum of a tensor along a given axis, it's sometimes useful to preserve the order of the tensor. This can be useful, for example, in broadcast scenarios.

Refer to section "2.1. Data Manipulation" for details on broadcasting.

```python
sum_G = ops.sum(G, dim=1, keepdim=True)
sum_G, sum_G.shape
```

This is useful, for example when generating a matrix whose rows sum up to 1. We can compute `G / sum_G` in this case.

```python
H = G / sum_G
H, ops.sum(H, dim=1)
```

We can also compute the cumulative sum of a tensor along a given axis. By design, it preserves the shape of the tensor.

```python
cumsum_G = ops.cumsum(G, axis=0)
cumsum_G
```
